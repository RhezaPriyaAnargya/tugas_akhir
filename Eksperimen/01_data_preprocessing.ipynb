```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from rdkit import Chem
from IPython.display import display
```

## Load Data

```python
BASE_DIR = Path("../dataset/dataset_raw")

files = {
    "EGFR": BASE_DIR / "EGFR_ChEMBL_raw.csv",
    "BRAF": BASE_DIR / "BRAF_ChEMBL_raw.csv"
}

for target, path in files.items():
    df = pd.read_csv(path)

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")
    print("Shape :", df.shape)
    print("Columns:")
    print(df.columns.tolist())
    print("\nMissing:")
    print(df.isna().sum())
    print("\nHead:")
    display(df.head())
```

```python
for target, path in files.items():
    df = pd.read_csv(path, low_memory=False)

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")

    for col in [
        "assay_type",
        "standard_type",
        "standard_relation",
        "standard_units"
    ]:
        print(f"\n{col}:")
        print(df[col].value_counts(dropna=False).head(10))
```

## Kurasi Data

```python
curated = {}

for target, path in files.items():
    df = pd.read_csv(path, low_memory=False)

    df = df[
        (df["assay_type"] == "B") &
        (df["standard_type"] == "IC50") &
        (df["standard_relation"] == "=") &
        (df["standard_units"] == "nM")
    ].copy()

    curated[target] = df

    print(f"{target}: {df.shape}")
    print(f"Missing SMILES   : {df['canonical_smiles'].isna().sum()}")
    print(f"Missing IC50     : {df['standard_value'].isna().sum()}")
    print(f"IC50 <= 0        : {(df['standard_value'] <= 0).sum()}")
```

```python
for target in curated:
    df = curated[target].copy()

    df = df.dropna(subset=["canonical_smiles", "standard_value"])
    df = df[df["standard_value"] > 0]
    df = df[~df["canonical_smiles"].str.contains(r"\.", regex=True)]

    curated[target] = df

    print(f"{target}: {df.shape}")
    print("Missing SMILES:", df["canonical_smiles"].isna().sum())
    print("Missing IC50  :", df["standard_value"].isna().sum())
    print("Multi-fragment:", df["canonical_smiles"].str.contains(r"\.").sum())
```

```python
for target in curated:
    df = curated[target].copy()

    mols = df["canonical_smiles"].apply(Chem.MolFromSmiles)
    valid = mols.notna()

    df = df.loc[valid].copy()
    df["mol"] = mols[valid]
    df["canonical_smiles"] = df["mol"].apply(
        lambda m: Chem.MolToSmiles(m, canonical=True)
    )

    curated[target] = df

    counts = df["canonical_smiles"].value_counts()
    multi = counts[counts > 1]

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")
    print(f"Valid records              : {len(df):,}")
    print(f"Unique compounds           : {counts.size:,}")
    print(f"Invalid SMILES removed     : {(~valid).sum():,}")
    print(f"Compounds with >1 measurement: {len(multi):,}")
    print(f"Extra measurement records  : {(multi - 1).sum():,}")
    print(f"Maximum measurements/compound: {counts.max()}")

    if len(multi) > 0:
        print("\nContoh senyawa dengan banyak pengukuran:")
        print(multi.head(10))
```

## Save data

```python
OUT_DIR = Path("../dataset/dataset_curated")
OUT_DIR.mkdir(parents=True, exist_ok=True)

for target, df in curated.items():
    export_df = df.drop(columns=["mol"], errors="ignore")
    path = OUT_DIR / f"{target}_curated_preaggregation.csv"
    export_df.to_csv(path, index=False)

    print(f"{target}: {len(export_df):,} records")
    print(f"Saved: {path}")
```

```python
for target, df in curated.items():
    df = df.copy()

    df["IC50_nM"] = pd.to_numeric(df["standard_value"], errors="coerce")
    df["pIC50"] = 9 - np.log10(df["IC50_nM"])

    curated[target] = df

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")
    print(f"Records   : {len(df):,}")
    print(f"IC50 min  : {df['IC50_nM'].min():.4f} nM")
    print(f"IC50 max  : {df['IC50_nM'].max():.4f} nM")
    print(f"pIC50 min : {df['pIC50'].min():.4f}")
    print(f"pIC50 max : {df['pIC50'].max():.4f}")

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    axes[0].hist(df["IC50_nM"], bins=50)
    axes[0].set_title("Before: IC50")
    axes[0].set_xlabel("IC50 (nM)")
    axes[0].set_ylabel("Frequency")

    axes[1].hist(df["pIC50"], bins=50)
    axes[1].set_title("After: pIC50")
    axes[1].set_xlabel("pIC50")
    axes[1].set_ylabel("Frequency")

    fig.suptitle(f"{target} — IC50 to pIC50 Transformation")
    plt.tight_layout()
    plt.show()
```

```python
for target, df in curated.items():
    print(f"\n{'='*75}")
    print(f"{target} — pIC50 DISTRIBUTION")
    print(f"{'='*75}")

    pic50 = df["pIC50"].dropna()

    print(f"Records        : {len(pic50):,}")
    print(f"Min            : {pic50.min():.4f}")
    print(f"Q1             : {pic50.quantile(0.25):.4f}")
    print(f"Median         : {pic50.median():.4f}")
    print(f"Mean           : {pic50.mean():.4f}")
    print(f"Q3             : {pic50.quantile(0.75):.4f}")
    print(f"Max            : {pic50.max():.4f}")

    print(f"\nOutside [3, 12] : {((pic50 < 3) | (pic50 > 12)).sum():,}")
    print(
        f"Inside [3, 12]  : "
        f"{((pic50 >= 3) & (pic50 <= 12)).sum():,}"
    )

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    axes[0].hist(
        df["IC50_nM"],
        bins=50,
        edgecolor="black"
    )
    axes[0].set_title(f"{target} — IC50")
    axes[0].set_xlabel("IC50 (nM)")
    axes[0].set_ylabel("Frequency")

    axes[1].hist(
        pic50,
        bins=50,
        edgecolor="black"
    )
    axes[1].set_title(f"{target} — pIC50")
    axes[1].set_xlabel("pIC50")
    axes[1].set_ylabel("Frequency")

    plt.tight_layout()
    plt.show()
```

## Penanganan Outlier 

```python
comparison = []

for target, df in curated.items():
    pic50 = df["pIC50"].dropna()

    # IQR
    q1 = pic50.quantile(0.25)
    q3 = pic50.quantile(0.75)
    iqr = q3 - q1

    iqr_lower = q1 - 1.5 * iqr
    iqr_upper = q3 + 1.5 * iqr

    iqr_outlier = (pic50 < iqr_lower) | (pic50 > iqr_upper)

    # Alshehade et al. (2026)
    range_outlier = (pic50 < 3) | (pic50 > 12)

    comparison.append({
        "Target": target,
        "Records": len(pic50),

        "IQR Q1": q1,
        "IQR Q3": q3,
        "IQR": iqr,
        "IQR Lower": iqr_lower,
        "IQR Upper": iqr_upper,
        "IQR Outliers": iqr_outlier.sum(),
        "IQR (%)": iqr_outlier.mean() * 100,

        "Range Lower": 3.0,
        "Range Upper": 12.0,
        "[3,12] Outliers": range_outlier.sum(),
        "[3,12] (%)": range_outlier.mean() * 100
    })

comparison_df = pd.DataFrame(comparison)

display(
    comparison_df.style
    .format({
        "IQR Q1": "{:.4f}",
        "IQR Q3": "{:.4f}",
        "IQR": "{:.4f}",
        "IQR Lower": "{:.4f}",
        "IQR Upper": "{:.4f}",
        "IQR (%)": "{:.3f}%",
        "Range Lower": "{:.1f}",
        "Range Upper": "{:.1f}",
        "[3,12] (%)": "{:.3f}%"
    })
)
```

memakai thresshold yang dipakai oleh alshehade kerena data yang harus dikeluarkan lebih sedikit dan berkemungkinan record tersebut erroneous

```python
filtered = {}

for target, df in curated.items():
    before = len(df)

    df = df[
        df["pIC50"].between(3, 12)
    ].copy()

    filtered[target] = df

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")
    print(f"Before : {before:,}")
    print(f"After  : {len(df):,}")
    print(f"Removed: {before - len(df):,}")
    print(f"Rate   : {(before - len(df)) / before * 100:.3f}%")
```

## Agregasi data

```python
for target, df in filtered.items():
    counts = df["canonical_smiles"].value_counts()
    multi = counts[counts > 1]

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")
    print(f"Records                    : {len(df):,}")
    print(f"Unique compounds           : {counts.size:,}")
    print(f"Compounds >1 measurement   : {len(multi):,}")
    print(f"Extra measurements         : {(multi - 1).sum():,}")
    print(f"Max measurements/compound  : {counts.max()}")
    print(f"Single-measurement compounds: {(counts == 1).sum():,}")
```

```python
for target, df in filtered.items():
    repeated = (
        df.groupby("canonical_smiles")
        .filter(lambda x: len(x) > 1)
    )

    comparison = (
        repeated.groupby("canonical_smiles")
        .agg(
            n_measurements=("pIC50", "size"),
            mean_pIC50=("pIC50", "mean"),
            median_pIC50=("pIC50", "median")
        )
        .reset_index()
    )

    comparison["abs_diff"] = (
        comparison["mean_pIC50"] -
        comparison["median_pIC50"]
    ).abs()

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")
    print(f"Repeated compounds : {len(comparison):,}")
    print(f"Mean |Mean-Median| : {comparison['abs_diff'].mean():.4f}")
    print(f"Median |Mean-Median|: {comparison['abs_diff'].median():.4f}")
    print(f"Max |Mean-Median|  : {comparison['abs_diff'].max():.4f}")
    print(
        f"Diff > 0.5 log unit: "
        f"{(comparison['abs_diff'] > 0.5).sum():,}"
    )
    print(
        f"Diff > 1.0 log unit: "
        f"{(comparison['abs_diff'] > 1.0).sum():,}"
    )
```

```python
for target, df in filtered.items():
    comparison = (
        df.groupby("canonical_smiles")
        .agg(
            n_measurements=("pIC50", "size"),
            mean_pIC50=("pIC50", "mean"),
            median_pIC50=("pIC50", "median")
        )
        .query("n_measurements > 1")
        .reset_index()
    )

    comparison["abs_diff"] = (
        comparison["mean_pIC50"] -
        comparison["median_pIC50"]
    ).abs()

    print(f"\n{target} — terbesar perbedaan Mean vs Median")
    display(
        comparison.sort_values("abs_diff", ascending=False)
        .head(10)
    )
```

Hasil analisis menunjukkan bahwa perbedaan mean dan median umumnya kecil. Pada EGFR, nilai absolut rata-rata perbedaan mean–median sebesar 0,0838 log unit, sedangkan median perbedaannya adalah 0,0000 log unit. Sebanyak 140 senyawa memiliki perbedaan lebih dari 0,5 log unit dan hanya 5 senyawa yang memiliki perbedaan lebih dari 1,0 log unit. Pada BRAF, perbedaan absolut rata-rata hanya 0,0194 log unit, dengan 10 senyawa memiliki perbedaan lebih dari 0,5 log unit dan hanya 1 senyawa lebih dari 1,0 log unit.

```python
aggregated = {}

for target, df in filtered.items():
    df_agg = (
        df.groupby("canonical_smiles", as_index=False)
        .agg(
            molecule_chembl_id=("molecule_chembl_id", "first"),
            pIC50=("pIC50", "median"),
            mean_pIC50=("pIC50", "mean"),
            n_measurements=("pIC50", "count")
        )
    )

    aggregated[target] = df_agg

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")
    print(f"Measurements before : {len(df):,}")
    print(f"Unique compounds    : {len(df_agg):,}")
    print(f"Single measurement  : {(df_agg['n_measurements'] == 1).sum():,}")
    print(f"Repeated measurement: {(df_agg['n_measurements'] > 1).sum():,}")
    print(f"Max measurements    : {df_agg['n_measurements'].max()}")
```

```python
for target, df in aggregated.items():
    duplicate_smiles = df["canonical_smiles"].duplicated().sum()
    missing_pic50 = df["pIC50"].isna().sum()
    outside_range = (~df["pIC50"].between(3, 12)).sum()

    print(f"\n{'='*60}")
    print(target)
    print(f"{'='*60}")
    print(f"Final compounds   : {len(df):,}")
    print(f"Duplicate SMILES  : {duplicate_smiles:,}")
    print(f"Missing pIC50     : {missing_pic50:,}")
    print(f"Outside [3,12]    : {outside_range:,}")
    print(f"pIC50 min         : {df['pIC50'].min():.4f}")
    print(f"pIC50 median      : {df['pIC50'].median():.4f}")
    print(f"pIC50 mean        : {df['pIC50'].mean():.4f}")
    print(f"pIC50 max         : {df['pIC50'].max():.4f}")

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.hist(df["pIC50"], bins=50, edgecolor="black")
    ax.set_title(f"{target} — Final pIC50 Distribution")
    ax.set_xlabel("pIC50")
    ax.set_ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
OUT_DIR = Path("../dataset/dataset_curated")
OUT_DIR.mkdir(parents=True, exist_ok=True)

for target, df in aggregated.items():
    path = OUT_DIR / f"{target}_curated.csv"
    df.to_csv(path, index=False)

    print(f"{target}: {len(df):,} compounds")
    print(f"Saved: {path}")
```
